```python
sales = [
12000, 15000, 18000, 10000,
22000, 25000, 14000, 16000,
20000, 17000, 13000
]
```

```python
total_sales = sum(sales)
average_sales = round(total_sales / len(sales), 2)
highest_sale = max(sales)
lowest_sale = min(sales)
number_of_sales = len(sales)
```

```python
print(f"Number of sales: {number_of_sales}")
print(f"Sales values: {sales}")
print(f"Total sales: ₹{total_sales}")
print(f"Average sales: ₹{average_sales}")
print(f"Highest sale: ₹{highest_sale}")
print(f"Lowest sale: ₹{lowest_sale}")
```
